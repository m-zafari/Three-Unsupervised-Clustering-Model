```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.datasets import make_moons
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering
from sklearn.cluster import DBSCAN
from sklearn.metrics.cluster import rand_score
from sklearn.metrics import jaccard_score
from sklearn.metrics import silhouette_score
from sklearn.metrics import davies_bouldin_score
import numpy as np
```

```python
def colour_detector(numeric_labels):
  colour = list()
  for label in numeric_labels:
    if label == 0:
      colour.append("red")
    elif label == 1:
      colour.append("blue")
    elif label == 2:
      colour.append("green")
    elif label == 3:
      colour.append("cyan")
    elif label == 4:
      colour.append("orange")
    elif label == 5:
      colour.append("olive")
    elif label == -1:        # noisy label
      colour.append("black")
    else:
      colour.append("yellow")

  colour = np.array(colour)
  return colour
```

```python
def xy_devider(xy):
  x, y = np.hsplit(xy, 2)
  x = [z[0] for z in x]
  y = [z[0] for z in y]
  return x, y
```

```python
circles_xy, circles_label = make_circles(n_samples= 1000, noise=0.08, random_state= 1, factor=0.5)
```

```python
x_circles, y_circles = xy_devider(circles_xy)
colour_circles = colour_detector(circles_label)
```

```python
plt.scatter(x_circles, y_circles)# c = colour_circles
plt.title("Noisy Circles")
plt.show
```

```python
circles_kmeans = KMeans(n_clusters=2, random_state= 1)
circles_kmeans.fit(circles_xy)
pred_circles_kmeans = circles_kmeans.labels_
```

```python
plt.scatter(x_circles, y_circles, c = colour_detector(pred_circles_kmeans))
plt.title("Predicted Label With Kmeans")
plt.show
```

```python
circles_clustering = AgglomerativeClustering(n_clusters=2)
circles_clustering.fit(circles_xy)
pred_circles_clustering = circles_clustering.labels_
```

```python
plt.scatter(x_circles, y_circles, c = colour_detector(pred_circles_clustering))
plt.title("AgglomerativeClustering on noisy circles")
plt.show
```

```python
circles_dbscan = DBSCAN(eps=0.12)
circles_dbscan.fit(circles_xy)
pred_circles_dbscan = circles_dbscan.labels_
```

```python
plt.scatter(x_circles, y_circles, c = colour_detector(pred_circles_dbscan))
plt.title("DBSCAN On Noisy Circles")
plt.show
```

```python
moons_xy, moons_label = make_moons(n_samples=1000, noise= 0.08, random_state=1)
```

```python
x_moons, y_moons = xy_devider(moons_xy)
```

```python
plt.scatter(x_moons, y_moons)
plt.title("Noisy Moons")
plt.show
```

```python
moons_kmeans = KMeans(n_clusters=2)
moons_kmeans.fit(moons_xy)
pred_moons_kmeans = moons_kmeans.labels_
```

```python

plt.scatter(x_moons, y_moons, c = colour_detector(pred_moons_kmeans))
plt.title("Predicted Label With KMeans")
plt.show
```

```python
moons_clustering = AgglomerativeClustering(n_clusters=2)
moons_clustering.fit(moons_xy)
pred_moons_clustering = moons_clustering.labels_
```

```python
plt.scatter(x_moons, y_moons, c = colour_detector(pred_moons_clustering))
plt.title("AgglomerativeClustering on noisy moons")
plt.show
```

```python
moons_dbscan = DBSCAN(eps=0.2)
moons_dbscan.fit(moons_xy)
pred_moons_dbscan = moons_dbscan.labels_
```

```python
plt.scatter(x_moons, y_moons, c = colour_detector(pred_moons_dbscan))
plt.title("DBSCAN on noisy moons")
plt.show
```

```python
blobs_xy, blobs_label = make_blobs(n_samples=1000, n_features=2, centers=5, random_state=1)
```

```python
x_blobs, y_blobs = xy_devider(blobs_xy)
colour_blobs = colour_detector(blobs_label)
```

```python
plt.scatter(x_blobs, y_blobs) # c = colour_blobs
plt.title("Isotropic Gassian Blobs")
plt.show
```

```python
blobs_kmeans = KMeans(n_clusters=5)
blobs_kmeans.fit(blobs_xy)
pred_blobs_kmeans = blobs_kmeans.labels_
```

```python
plt.scatter(x_blobs, y_blobs, c = colour_detector(pred_blobs_kmeans))
plt.title("kmeans on blobs")
plt.show
```

```python
blobs_clustering = AgglomerativeClustering(n_clusters=5)
blobs_clustering.fit(blobs_xy)
pred_blobs_clustering = blobs_clustering.labels_
```

```python
plt.scatter(x_blobs, y_blobs, c = colour_detector(pred_blobs_clustering))
plt.title("AgglomerativeClustering on blobs")
plt.show
```

```python
blobs_dbscan = DBSCAN(eps=0.5)
blobs_dbscan.fit(blobs_xy)
pred_blobs_dbscan = blobs_dbscan.labels_
```

```python
plt.scatter(x_blobs, y_blobs, c = colour_detector(pred_blobs_dbscan))
plt.title("DBSCAN on blobs")
plt.show
```

```python
# KMeans scores

# rand score
circle_kmeans_randscore = rand_score(circles_label, pred_circles_kmeans)
blobs_kmeans_randscore = rand_score(blobs_label, pred_blobs_kmeans)
moons_kmeans_randscore = rand_score(moons_label, pred_moons_kmeans)

# jacard score
circle_kmeans_jacardscore = jaccard_score(circles_label, pred_circles_kmeans)
blobs_kmeans_jacardscore = jaccard_score(blobs_label, pred_blobs_kmeans, average = "micro")
moons_kmeans_jacardscore = jaccard_score(moons_label, pred_moons_kmeans)

# silhouette score
circle_kmeans_silhouettescore = silhouette_score(circles_xy, pred_circles_kmeans)
blobs_kmeans_silhouettescore = silhouette_score(blobs_xy, pred_blobs_kmeans)
moons_kmeans_silhouettescore = silhouette_score(moons_xy, pred_moons_kmeans)

# dunn score
circle_kmeans_dunnscore = davies_bouldin_score(circles_xy, pred_circles_kmeans)
blobs_kmeans_dunnscore = davies_bouldin_score(blobs_xy, pred_blobs_kmeans)
moons_kmeans_dunnscore = davies_bouldin_score(moons_xy, pred_moons_kmeans)
```

```python
# AgglomerativeClustering scores

# rand score
circle_clustering_randscore = rand_score(circles_label, pred_circles_clustering)
blobs_clustering_randscore = rand_score(blobs_label, pred_blobs_clustering)
moons_clustering_randscore = rand_score(moons_label, pred_moons_clustering)

# jacard score
circle_clustering_jacardscore = jaccard_score(circles_label, pred_circles_clustering)
blobs_clustering_jacardscore = jaccard_score(blobs_label, pred_blobs_clustering, average = "micro")
moons_clustering_jacardscore = jaccard_score(moons_label, pred_moons_clustering)

# silhouette score
circle_clustering_silhouettescore = silhouette_score(circles_xy, pred_circles_clustering)
blobs_clustering_silhouettescore = silhouette_score(blobs_xy, pred_blobs_clustering)
moons_clustering_silhouettescore = silhouette_score(moons_xy, pred_moons_clustering)

# dunn score
circle_clustering_dunnscore = davies_bouldin_score(circles_xy, pred_circles_clustering)
blobs_clustering_dunnscore = davies_bouldin_score(blobs_xy, pred_blobs_clustering)
moons_clustering_dunnscore = davies_bouldin_score(moons_xy, pred_moons_clustering)
```

```python
# DBSCAN scores

# rand score
circle_dbscan_randscore = rand_score(circles_label, pred_circles_dbscan)
blobs_dbscan_randscore = rand_score(blobs_label, pred_blobs_dbscan)
moons_dbscan_randscore = rand_score(moons_label, pred_moons_dbscan)

# jacard score
circle_dbscan_jacardscore = jaccard_score(circles_label, pred_circles_dbscan, average =  "micro")
blobs_dbscan_jacardscore = jaccard_score(blobs_label, pred_blobs_dbscan, average = "micro")
moons_dbscan_jacardscore = jaccard_score(moons_label, pred_moons_dbscan, average = "micro")

# silhouette score
circle_dbscan_silhouettescore = silhouette_score(circles_xy, pred_circles_dbscan)
blobs_dbscan_silhouettescore = silhouette_score(blobs_xy, pred_blobs_dbscan)
moons_dbscan_silhouettescore = silhouette_score(moons_xy, pred_moons_dbscan)

# dunn score
circle_dbscan_dunnscore = davies_bouldin_score(circles_xy, pred_circles_dbscan)
blobs_dbscan_dunnscore = davies_bouldin_score(blobs_xy, pred_blobs_dbscan)
moons_dbscan_dunnscore = davies_bouldin_score(moons_xy, pred_moons_dbscan)
```
